```python
by_breed = (
    dogs.groupby("breed", as_index=False)
    .agg(
        n=("id", "size"),
        mean_mm3=("brain_volume_mm3", "mean"),
        min_mm3=("brain_volume_mm3", "min"),
        max_mm3=("brain_volume_mm3", "max"),
    )
    .sort_values("mean_mm3", ascending=False)
    .reset_index(drop=True)
)
by_breed.round({"mean_mm3": 1, "min_mm3": 1, "max_mm3": 1})
```

```python
%matplotlib inline
import matplotlib.colors as mcolors
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

order = by_breed.sort_values("mean_mm3")["breed"].tolist()
means = by_breed.set_index("breed").loc[order, "mean_mm3"]
positions = {breed: i for i, breed in enumerate(order)}
breed_purpose = dogs.drop_duplicates("breed").set_index("breed")["purpose"]

# Legend follows the chart from the largest mean (top) downward.
purpose_order = []
for breed in reversed(order):
    purpose = breed_purpose.loc[breed]
    if purpose not in purpose_order:
        purpose_order.append(purpose)

# Distinct color for each purpose title. Hues are stepped so neighbors in the legend stay apart.
purpose_colors = {
    purpose: mcolors.hsv_to_rgb([((i * 0.618033988749895) % 1), 0.62, 0.93])
    for i, purpose in enumerate(purpose_order)
}
bar_colors = [purpose_colors[breed_purpose.loc[breed]] for breed in order]

fig, ax = plt.subplots(figsize=(16, 10), layout="constrained")
ax.barh(
    range(len(order)),
    means,
    color=bar_colors,
    zorder=1,
    edgecolor="white",
    linewidth=0.6,
)
ax.scatter(
    dogs["brain_volume_mm3"],
    dogs["breed"].map(positions),
    color="black",
    s=32,
    zorder=2,
)
ax.set_yticks(range(len(order)))
ax.set_yticklabels(order)
ax.set_xlabel("Brain volume (mm³)")
ax.set_ylabel("Breed")
ax.set_title("Brain volume by dog breed")
ax.set_xlim(0, dogs["brain_volume_mm3"].max() * 1.05)

handles = [
    Patch(
        facecolor=purpose_colors[purpose],
        edgecolor="#666666",
        linewidth=0.4,
        label=purpose,
    )
    for purpose in purpose_order
]
handles.append(
    Line2D(
        [0],
        [0],
        marker="o",
        linestyle="None",
        markerfacecolor="black",
        markeredgecolor="black",
        markersize=6,
        label="Individual dog",
    )
)
ax.legend(
    handles=handles,
    title="Purpose",
    loc="center left",
    bbox_to_anchor=(1.02, 0.5),
    frameon=False,
    fontsize=8,
    title_fontsize=9,
)
plt.show()
```
